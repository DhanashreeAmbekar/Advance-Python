```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df=pd.read_csv("C://Users//DELL//Desktop//Excel_Data//website data.csv")
```

```python
df.head()
```

```python
df.columns=df.iloc[0]
df=df.drop(index= 0).reset_index(drop=True)
df.columns=["channel group","datehour","users","sessions","engaged sessions"," average engagement time per session", "engaged session per user","events per session","engagement rate","event count"]         
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df["datehour"]=pd.to_datetime(df["datehour"],format="%Y%m%d%H",errors="coerce")
df.head()
```

```python
df.info()
```

```python
df['hour']=df['datehour'].dt.hour
```

```python
df.head()
```

```python
numeric_cols=df.columns.drop(["channel group","datehour","hour"])
df[numeric_cols]=df[numeric_cols].apply(pd.to_numeric,errors='coerce')
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df.describe()
```

### Data Visualisation

### Sessions and users over time

```python
sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(10,5))
df.groupby('datehour')[['sessions','users']].sum().plot(ax=plt.gca())
plt.title("sessions and users overtime")
plt.xlabel("datehour")
plt.ylabel("count")
plt.show()
```

### total users by channel

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df,x="channel group",y="users",estimator=np.sum)
plt.title("total users by channel")
plt.show()
```

```python
df.head(3)
```

### avg event per session by channel

```python
df['channel group'].value_counts()
```

```python
plt.figure(figsize=(12,8))
sns.barplot(data=df,x="channel group",y="events per session",estimator=np.mean)
plt.show()
```

### how engagement rate vary across different channel ?

```python
plt.figure(figsize=(12,5))
sns.boxplot(data=df,x="channel group", y="engagement rate")
plt.show()
```

### Engaged vs Non engaged sessions

```python
df['non engaged sessions']=df['sessions']-df['engaged sessions']
```

```python
df['non engaged sessions']
```

```python
df.head()
```

```python
plt.figure(figsize=(10,5))
df.groupby('channel group')[['engaged sessions','non engaged sessions']].sum().plot(ax=plt.gca())
plt.title("Engaged vs Non engaged sessions")
plt.xlabel("channel group")
plt.ylabel("sessions count")
plt.show()
```

### Traffic by hour and channel

```python
heatmap_data=df.groupby(['channel group','hour'])['sessions'].sum().unstack().fillna(0)

plt.figure(figsize=(15,8))
sns.heatmap(heatmap_data , linewidths=5, annot=True,fmt='.0f')
plt.show()
```

###  engagement rate vs sessions over time

```python
df_plot=df.groupby("datehour")[["engagement rate","sessions"]].mean().reset_index()

plt.figure(figsize=(15,5))
plt.plot(df_plot["datehour"],df_plot["engagement rate"])
plt.plot(df_plot["datehour"],df_plot["sessions"])
plt.show()

                   
```
