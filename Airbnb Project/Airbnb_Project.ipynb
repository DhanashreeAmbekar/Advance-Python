### handeling missing values

```python
df["last review"]=pd.to_datetime(df["last review"],errors="coerce")
```

```python
df.info()
```

```python
df["reviews per month"].fillna('0',inplace=True)
```

```python
df["last review"].fillna(df['last review'].min(),inplace=True)
```

```python
df.info()
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df.dropna(subset=["NAME","host name"],inplace=True)
```

```python
df.isna().sum()
```

```python
df=df.drop(columns=["license","house_rules"],errors="ignore")
```

```python
df.head()
```

```python
df.shape
```

```python
df["price"]=df["price"].replace('[\$,]', '', regex=True)
```

```python
df["price"]=df["price"].astype(float)
```

```python
df["service fee"]=df["service fee"].replace('[\$,]','',regex=True)
```

```python
df["service fee"]=df["service fee"].astype(float)
```

### Data Visualisation

### what is the distribution of listing prices

```python
plt.figure(figsize=(8,5))
sns.histplot(df["price"],kde=True,bins=50)
plt.title("Distribution of listing price")
plt.xlabel("price")
plt.ylabel("frequncy")
plt.show()
```

### how are different room types distributed ?

```python
df["room type"].value_counts()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x="room type",data=df)
plt.title("room type distribution")
plt.xlabel("room type")
plt.ylabel("count")
plt.show()
```

### how are listings distributed across different neighborhoods ?

```python
df["neighbourhood"].value_counts()
```

```python
plt.figure(figsize=(12,8))
sns.countplot(y="neighbourhood group", data=df, order=df["neighbourhood group"].value_counts().index)
plt.show()
```

### what is relationship between price and room type ?

```python
plt.figure(figsize=(10,5))
sns.boxplot(x="room type", y="price", hue="room type",data=df)
plt.title("price vs room type")
plt.show()
```

### how review change over time ?

```python
df.head(3)
```

```python
df.info()
```

```python
reviews_over_time=df.groupby(df["last review"].dt.to_period('M')).size()
```

```python
reviews_over_time
```

```python
plt.figure(figsize=(8,5))
reviews_over_time.plot(kind='line',color='red')
plt.title("Reviews over time")
plt.xlabel("Date")
plt.ylabel("Number of Reviews")
plt.show()
```
